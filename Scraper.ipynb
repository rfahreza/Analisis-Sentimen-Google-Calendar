```python
!pip install google-play-scraper
# Mengimpor pustaka google_play_scraper untuk mengakses ulasan dan informasi aplikasi dari Google Play Store.
from google_play_scraper import app, reviews, Sort, reviews_all

#Mengimpor csv untuk menyimpan hasil scrapping ke dalam file csv
import csv
#Mengimpor pandas untuk membuat data frame
import pandas as pd
```

# Scrapping

```python
scrap_review, _ = reviews(
                'com.google.android.calendar',
                lang='id',
                country='id',
                sort=Sort.MOST_RELEVANT,
                count=10000
)
```

Saya menggunakan ulasan dari aplikasi **Google Calendar** dengan mengambil **10.000** ulasan pengguna yang dianggap paling relevan.

```python
# Mengubah hasil scraping menjadi dataframe
df = pd.DataFrame(scrap_review)
df.head()
```

Membuat data frame baru yang akan digunakan untuk analisis agar lebih storage lebih efisien, fokus, dan juga praktis

```python
# memilih baris yang akan digunakan untuk analisis
data_review = df[['content', 'score']]
data_review.head()
```

# Save file to csv

```python
data_review.to_csv('google_calendar_review.csv', index=False)
```
